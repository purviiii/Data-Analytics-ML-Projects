# **Model Building**

```python
x=df.iloc[:,:-1]
y=df['Churn']
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
x_train
```

```python
y_train
```

```python
y_train.value_counts()
```

```python
x_test
```

```python
y_test
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model=RandomForestClassifier()
```

```python
#Train the Model
model.fit(x_train,y_train)
```

```python
#Test the Model
y_pred=model.predict(x_test)
```

```python
#Evaluate the model
from sklearn.metrics import *
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
precision_score(y_test,y_pred)
```

```python
recall_score(y_test,y_pred)
```

```python
param_grid={
    'n_estimators':[100,200,300,400],
    'max_depth':[1,5,10,12],
    'min_samples_split':[2,5,7,9],
    'min_samples_leaf':[1,2,4,6],
    'criterion':['gini','entropy']
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
gv=GridSearchCV(estimator=model,param_grid=param_grid,cv=3)
```

```python
gv.fit(x_train,y_train)
```

```python
gv.best_params_
```

```python
best_model = gv.best_estimator_
y_pred_tuned = best_model.predict(x_test)

print("Tuned Accuracy:", round(accuracy_score(y_test, y_pred_tuned) * 100, 2), "%")
print("Tuned Precision:", round(precision_score(y_test, y_pred_tuned) * 100, 2), "%")
print("Tuned Recall:", round(recall_score(y_test, y_pred_tuned) * 100, 2), "%")
print("ROC-AUC Score:", round(roc_auc_score(y_test, best_model.predict_proba(x_test)[:, 1]), 3))
```

```python
importances = pd.Series(best_model.feature_importances_, index=x.columns).sort_values(ascending=False)

plt.figure(figsize=(7, 4))
sns.barplot(x=importances.head(10), y=importances.head(10).index, palette='viridis')
plt.title('Top 10 Features Driving Customer Churn')
plt.xlabel('Feature Importance Score')
plt.ylabel('Features')
plt.show()
```

### Key Business Insights & Retention Recommendations

* **Model Performance:** Hyperparameter tuning via 3-fold GridSearchCV improved test accuracy to **79.46%** with a **0.826 ROC-AUC score** (`max_depth=10`, `min_samples_leaf=2`, `min_samples_split=7`, `n_estimators=100`).
* **Top Churn Drivers:** Contract type (`Contract`), customer tenure (`tenure`), monthly charges (`MonthlyCharges`), and total charges (`TotalCharges`) are the strongest predictors of customer churn.
* **Actionable Strategy:** Target month-to-month subscribers with high monthly bills during their first 6–12 months using discounted annual contract upgrades and bundled tech/security add-ons.
